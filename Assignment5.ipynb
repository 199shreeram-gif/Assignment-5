```python
import pandas as pd
# Explicitly parse the semicolon-separated CSV file
df = pd.read_csv('/content/student-mat.csv', sep=';')
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select only numerical columns for covariance and correlation
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Calculate Covariance Matrix
cov_matrix = numeric_df.cov()
print('--- Covariance Matrix ---')
display(cov_matrix)

# Calculate Correlation Matrix
corr_matrix = numeric_df.corr()
print('\n--- Correlation Matrix ---')
display(corr_matrix)
```

```python
# Plot the correlation matrix heatmap for better visualization
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', square=True, linewidths=0.5)
plt.title('Correlation Matrix of Student Numerical Features')
plt.show()
```
